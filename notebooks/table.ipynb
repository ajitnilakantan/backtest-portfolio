```python
import pandas as pd
from IPython.display import HTML, Javascript
from itables import init_notebook_mode, to_html_datatable
from tabulate import tabulate

# 1. Create Sample DataFrame
data = {
    "Long Header Number One": ["Very long text one", "Very long text two"],
    "Long Header Number Two": ["Some data here", "More random text here"],
}
df = pd.DataFrame(data)

# 2. Format Text in Cells First (using original column names)
df["Long Header Number One"] = df["Long Header Number One"].str.capitalize()

# 3. Create a temporary copy for wrapping headers (keeps original DataFrame clean)
display_df = df.copy()
display_df.columns = [col.replace(" ", "\n") for col in display_df.columns]

# 4. Print with Tabulate
# print(tabulate(display_df, headers='keys', tablefmt='grid', showindex=False))

# Turn on interactive mode globally
init_notebook_mode(all_interactive=True)

html = to_html_datatable(df, display_logo_when_loading=True)
display({"text/html": html}, raw=True)
display({"text/html": HTML(html).data}, raw=True)
```

```python
import pandas as pd

# 1. Create Sample DataFrame with all requested data types
data = {
    "Long Header Number One": ["Very long text one", "Very long text two"],
    "Large Count Quantity": [1250500, 87400],
    "Total Revenue Amount": [45320.50, 120500.75],
    "Average Score Metric": [85.1234, 92.5678],
    "Growth Rate Percentage": [0.1234, -0.0456],
}
df = pd.DataFrame(data)

# 2. Format Text & Numeric Columns (Original names remain intact)
df["Long Header Number One"] = df["Long Header Number One"].str.capitalize()

# Thousands Separator (Integer or float)
df["Large Count Quantity"] = df["Large Count Quantity"].map("{:,}".format)

# Currency Formatting ($ + Thousands + 2 Decimals)
df["Total Revenue Amount"] = df["Total Revenue Amount"].map("${:,.2f}".format)

# Numbers to Exactly 2 Decimal Places
df["Average Score Metric"] = df["Average Score Metric"].map("{:.2f}".format)

# Percentages to Exactly 1 Decimal Place (Multiplies by 100 and adds %)
df["Growth Rate Percentage"] = df["Growth Rate Percentage"].map("{:.1%}".format)

# 3. Create a temporary copy for wrapping headers
display_df = df.copy()
display_df.columns = [col.replace(" ", "\n") for col in display_df.columns]

# 4. Print with Tabulate
print(tabulate(display_df, headers="keys", tablefmt="grid", showindex=False))
# Display in Jupyter Notebook or export to HTML
with pd.option_context("display.notebook_repr_html", True):
    display_html(display_df.to_html(), raw=True)

# pd.set_option('styler.render.max_elements', 9999999)
# display(display_df)
```
